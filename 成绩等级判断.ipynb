```python
score = float(input('请输入您的成绩：'))
if 90<= score <= 100:
    print(f'成绩{score:.2f}为优秀！')
elif 80 <= score <90:
    print(f'成绩{score:.2f}为良好！')
elif 70 <= score:
    print(f'成绩{score:.2f}为中等！')
elif 60 <= score:
    print(f'成绩{score:.2f}为及格！')
elif 0 <= score:
    print(f'成绩{score:.2f}为不及格！')
else:
    print('您输入的成绩不合法')
'''
要把不合法的情况往前提，首先判断合不合法，不然就会有遗漏，这里输入101，就会显示中等'''
score = float(input('请输入您的成绩：'))
if score >100 or score < 0:
    print('您输入的成绩不合法')    
elif 90<= score <= 100:
    print(f'成绩{score:.2f}为优秀！')
elif 80 <= score <90:
    print(f'成绩{score:.2f}为良好！')
elif 70 <= score:
    print(f'成绩{score:.2f}为中等！')
elif 60 <= score:
    print(f'成绩{score:.2f}为及格！')
elif 0 <= score:
    print(f'成绩{score:.2f}为不及格！')
```

要点：

1. 从上到下依次判断，命中一个就跳出，后面的 elif 不再看

2. elif 隐含“前面所有条件都为假”，所以第二个分支开始不用再写下界

3. 顺序不能乱：先写 >= 90，再写 >= 80；反过来 95 分会被判成 B

4. 最后一个分支用 else 兜底，保证任何输入都有输出

5. elif 可以写多个，else 最多一个，且必须在最后

6. elif 不是 else if，Python 没有 else if 这种写法

7. else 后面没有条件，直接跟冒号

8. else 必须和 if 对齐
