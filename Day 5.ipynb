1. range()：生成数字序列

|格式|生成|实例|
|-------|-----|------|
|range(n) |0 到 n-1|range(5) → 0,1,2,3,4|
|range(start, stop)	|start 到 stop-1|	range(1, 6) → 1,2,3,4,5|
|range(start, stop, step)|从 start 到 stop-1，步长 step|range(1, 10, 2) → 1,3,5,7,9|

2. 
包含起点，不包含终点（左闭右开）

range(5) 不是 1 到 5，是 0 到 4

步长可以是负数：range(10, 0, -2) → 10,8,6,4,2

步长不能为 0，会报 ValueError

    while 条件:
        循环体
条件为 True 就执行循环体，执行完再判断

条件为 False 时退出

循环体内必须有能让条件最终变假的代码，否则死循环

```python
for i in range(1,10):
    for j in range(1,i+1):
        print(f'{i} * {j} = {i*j}',end='   ')       
    print('\n')

```

```python
import random
answer = random.randint(1,100)
print('请猜一个数字，范围为1-100')
guess = int(input('请输入一个整数，1-100范围内'))
left = 1
right = 100
while guess != answer:
    if guess >= right or guess <= left:
        guess = int(input('不合法输入，请重新输入'))
    elif guess < answer:
        left = guess
        print(f'猜小了,请输入{left}-{right}')
    elif guess > answer:
        right = guess
        print(f'猜大了,请输入{left}-{right}')
    guess = int(input())
print('猜对了')
```
